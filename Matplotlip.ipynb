```python
import numpy as np
import matplotlib.pyplot as plt
```

# Basic line Plot

```python
x = np.array([1, 2, 3, 4])
y = np.array([10, 30, 25, 50])

plt.plot(x, y)
plt.title("Basic Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

# Custom Style

```python
x = np.array([1, 2, 3, 4, 6])
y = np.array([2, 7, 6, 8, 15])

plt.plot(y, linestyle='dotted', color='green', marker='o')
plt.title("Styled Line Example")
plt.show()
```

# Multiple Lines

```python
x = np.array([1, 2, 3, 4, 6])
y1 = np.array([3, 8, 1, 10, 15])
y2 = np.array([4, 9, 3, 12, 18])
y3 = np.array([4, 7, 9, 16, 16])

plt.plot(x, y1, label="Line 1", color="red", marker="x")
plt.plot(x, y2, label="Line 2", color="green", marker="s", linestyle="--")
plt.plot(x, y3, label="Line 3", color="blue", marker="o", linestyle="-")

plt.title("Multiple Lines")
plt.xlabel('X label ')
plt.ylabel('Y label ')
plt.legend()
plt.show()
```

# Bar CHART

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array(['Node.js', 'Java', 'C++', 'TypeScript'])
y = np.array([42, 22, 8, 33])

plt.bar(x, y, color='orange')
plt.title('Bar Chart Example')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.show()
```

# Scatter Plot

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([10, 12, 32, 14, 25])
y = np.array([5, 4, 6, 3, 7])

plt.scatter(x, y, color='RED', s=50)
plt.title('Scatter Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

# Histogram

```python
data = np.array([3, 4, 5, 6, 8, 13, 13, 15, 19])

plt.hist(data, bins=7, color='skyblue', edgecolor='black')
plt.title("Histogram Example")
plt.xlabel('Values')
plt.ylabel('Count')
plt.show()
```

# SubPlot

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4])
y1 = np.array([2, 4, 8, 23])
y2 = np.array([3, 16, 19, 22])

plt.figure(figsize=(8, 4))  # width, height

plt.subplot(1, 2, 1)
plt.plot(x, y1, color='red')
plt.title('Plot 1')

plt.subplot(1, 2, 2)
plt.bar(x, y2, color='orange')
plt.title('Plot 2')

plt.tight_layout()
plt.show()
```

# SAVE

```python
x = np.array([12, 22, 23, 24])
y = np.array([14, 18, 16, 10])

plt.plot(x, y, color='green')
plt.title('Save Figure Example')
plt.savefig('my_plot.png')
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

# Make data
u = np.linspace(0, 2 * np.pi, 100)
v = np.linspace(0, np.pi, 100)
x = 10 * np.outer(np.cos(u), np.sin(v))
y = 10 * np.outer(np.sin(u), np.sin(v))
z = 10 * np.outer(np.ones(np.size(u)), np.cos(v))

# Plot the surface
ax.plot_surface(x, y, z)

# Set an equal aspect ratio
ax.set_aspect('equal')

plt.show()
```
