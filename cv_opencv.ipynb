```python
import cv2
```

```python
from google.colab.patches import cv2_imshow
image = cv2.imread("/content/cat_img.png")
cv2_imshow(image)
```

```python
print('SHAPE', image.shape)
# 1254 ==> height
# 1254 ==> width
# 3 ==> color
print('height', image.shape[0])
print('width', image.shape[1])
print('CHANNELS', image.shape[2])
```

```python
(b , g, r) = image[100, 50]
print(b ,g, r )
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
resized = cv2.resize(gray, (300, 300))
cv2_imshow(resized)
```

```python
cropped = gray[50:350, 100:200]
cv2_imshow(cropped)
```

```python
rotate = cv2.rotate(gray, cv2.ROTATE_90_CLOCKWISE)
cv2_imshow(rotate)
```

```python
flippet = cv2.flip(gray,1)
cv2_imshow(flippet)
```

```python
cv2.imwrite("flippet.cat.jpg", flippet)
```
